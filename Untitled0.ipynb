```python
price = int(input("상품 가격(원): "))   # 입력: 숫자를 받아 price에 저장
total = price * 1.1                    # 처리: 부가세 10%를 더한 금액 계산
print("결제 금액:", int(total), "원")   # 출력: 소수점을 버리고 화면에 표시
```

```python
onion = int(input("양파 가격(원): "))
tomato = int(input("토마토 가격(원): "))
potato = int(input("감자 가격(원): "))

print("\n" + "="*20)
print(f"      ★ 메뉴판 ★")
print("-"*20)
print(f" 양파   :  {int(onion*1.1):>5,}원")
print(f" 토마토  :  {int(tomato*1.1):>5,}원")
print(f" 감자   :  {int(potato*1.1):>5,}원")
print("="*20)

print("\n원하시는 구매 수량을 적어주세요 ")
conion = int(input("양파(개) : "))
ctomato = int(input("토마토(개) : "))
cpotato = int(input("감자(개) : "))

print("\n" + "="*22)
print(f"      ★ 총 가격 ★")
print("-"*22)
print(f"   {int(onion*1.1*conion + tomato*1.1*ctomato + potato*1.1*cpotato):>5,}원 입니다.")
print("="*22)
```
